```python
import re

clean = df.copy()
report = [("Shuru", len(clean))]

def step(name, mask_keep):
    global clean
    before = len(clean)
    clean = clean[mask_keep(clean)].copy()
    report.append((name, before - len(clean)))

# 1. Duplicate content, phir duplicate title (pehli row rakho)
step("Duplicate content hataye", lambda d: ~d.duplicated(subset="content", keep="first"))
step("Duplicate title hataye",   lambda d: ~d.duplicated(subset="title", keep="first"))

# 2. Puzzle-hint articles
puzzle_pat = r"Wordle|Connections|Strands|Mini Crossword|Spelling Bee"
step("Puzzle articles hataye", lambda d: ~d["title"].str.contains(puzzle_pat, case=False, regex=True))

# 3. 5000+ words (live blogs, transcripts, deal trackers)
step("5000+ words hataye", lambda d: d["content_words"] <= 5000)

# 4. Titles ke extra spaces
clean["title"] = clean["title"].str.replace(r"\s+", " ", regex=True).str.strip()

# 5. content_clean: HTML, URLs, emails, extra spaces
def clean_text(t):
    t = re.sub(r"<[^>]+>", " ", t)
    t = re.sub(r"http\S+|www\.\S+", " ", t)
    t = re.sub(r"\S+@\S+\.\S+", " ", t)
    t = re.sub(r"&nbsp;|&amp;", " ", t)
    return re.sub(r"\s+", " ", t).strip()

clean["content_clean"] = clean["content"].apply(clean_text)
clean["content_words"] = clean["content_clean"].str.split().str.len()

# 6. Safai ke baad jo bohat chhote reh gaye
step("Safai ke baad 50 words se kam hataye", lambda d: d["content_words"] >= 50)

clean = clean.reset_index(drop=True)

print("Cleaning report:")
for name, n in report:
    print(f"  {name}: {n}")
print("Bache:", len(clean))
print(clean["source"].value_counts().to_string())
```

```python
import glob, re, html, json, unicodedata
from urllib.parse import urlparse
import numpy as np
import pandas as pd

RAW_GLOB = "../Task-1-Web-Scraping-100k/dataset/*.zip"
df_raw = pd.concat(
    [pd.read_csv(f) for f in sorted(glob.glob(RAW_GLOB))],
    ignore_index=True,
)
print("Raw shape:", df_raw.shape)
print("Columns:", df_raw.columns.tolist())

expected = ["title", "description", "date_of_publication",
            "author_name", "content", "url", "source"]
print("Missing columns:", [c for c in expected if c not in df_raw.columns])

work = df_raw.copy(deep=True)
work["row_id"] = range(len(work))
for c in expected:
    work[c + "_raw"] = work[c]
print("Working copy ready:", work.shape)
```

```python
MISSING_TOKENS = {"", "n/a", "na", "null", "none", "nan", "-", "--"}

def normalize_missing(series, extra=()):
    s = series.astype(object)
    def f(v):
        if pd.isna(v):
            return pd.NA
        if isinstance(v, str):
            t = v.replace("\u00a0", " ").strip()
            if t.casefold() in MISSING_TOKENS or t.casefold() in extra:
                return pd.NA
            return t
        return v
    return s.map(f)

before = work[["title", "description", "author_name", "content", "url"]].isna().sum()

for c in ["title", "description", "content", "url", "date_of_publication"]:
    work[c] = normalize_missing(work[c])

# "Unknown" sirf author mein missing maana jayega
work["author_name"] = normalize_missing(work["author_name"], extra={"unknown"})

after = work[["title", "description", "author_name", "content", "url"]].isna().sum()
print(pd.DataFrame({"null_before": before, "null_after": after}))
print("\nauthor_name 'Unknown' bacha:", (work["author_name"] == "Unknown").sum())
```

```python
from html.parser import HTMLParser

class TextExtractor(HTMLParser):
    SKIP = {"script", "style", "noscript"}
    def __init__(self):
        super().__init__(convert_charrefs=True)
        self.parts, self._skip = [], 0
    def handle_starttag(self, tag, attrs):
        if tag in self.SKIP:
            self._skip += 1
    def handle_endtag(self, tag):
        if tag in self.SKIP and self._skip:
            self._skip -= 1
    def handle_data(self, data):
        if self._skip == 0 and data and data.strip():
            self.parts.append(data.strip())

INVISIBLE = dict.fromkeys(map(ord, "\u200b\u200c\u200d\u2060\ufeff\u00ad"), None)
HAS_TAG = re.compile(r"<[a-zA-Z/!][^>]*>")

def clean_text(v):
    if pd.isna(v):
        return pd.NA
    t = str(v)
    if HAS_TAG.search(t):                      # parser sirf tab jab HTML ho (tez)
        p = TextExtractor()
        try:
            p.feed(t); p.close()
            t = " ".join(p.parts)
        except Exception:
            t = re.sub(r"<[^>]+>", " ", t)
    t = html.unescape(t)
    t = t.translate(INVISIBLE)
    t = unicodedata.normalize("NFKC", t)
    t = (t.replace("\u201c", '"').replace("\u201d", '"')
          .replace("\u2018", "'").replace("\u2019", "'"))
    t = re.sub(r"\s+", " ", t).strip()
    return t if t else pd.NA

for c in ["title", "description", "author_name", "content"]:
    work[c + "_clean"] = work[c].map(clean_text)

print(work[["title_clean", "description_clean", "author_name_clean", "content_clean"]].isna().sum())

changed = (work["content"].astype(str) != work["content_clean"].astype(str)).sum()
print("\ncontent jis mein tabdili hui:", changed)
print("content_clean mein new line:", work["content_clean"].astype(str).str.contains("\n").sum())
print("description_clean mein new line:", work["description_clean"].astype(str).str.contains("\n").sum())
```

```python
# --- URL validity aur domain ---
def valid_url(v):
    if pd.isna(v):
        return False
    try:
        p = urlparse(str(v).strip())
        return p.scheme.lower() in {"http", "https"} and bool(p.hostname) and "." in p.hostname
    except Exception:
        return False

work["url_valid"] = work["url"].map(valid_url)
work["domain"] = work["url"].map(
    lambda v: urlparse(str(v)).hostname.lower() if valid_url(v) else pd.NA
)

# --- Dates ---
work["published_at_utc"] = pd.to_datetime(
    work["date_of_publication"], errors="coerce", utc=True, format="mixed"
)
work["published_year"] = work["published_at_utc"].dt.year
work["published_month"] = work["published_at_utc"].dt.month
work["published_weekday"] = work["published_at_utc"].dt.day_name()

# --- Length features ---
def wc(s):
    return s.fillna("").astype(str).str.findall(r"\b[\w'-]+\b").str.len()

work["title_char_count"] = work["title_clean"].fillna("").astype(str).str.len()
work["content_char_count"] = work["content_clean"].fillna("").astype(str).str.len()
work["content_word_count"] = wc(work["content_clean"])
work["title_too_short"] = work["title_char_count"] < 8
work["content_too_short"] = work["content_char_count"] < 200

# --- Block / paywall check ---
block = re.compile(r"access denied|enable javascript|captcha|robot check|page not found|403 forbidden|subscribe to continue", re.I)
work["possible_block_or_paywall"] = work["content_clean"].fillna("").astype(str).str.contains(block)

print("URL invalid:", (~work["url_valid"]).sum())
print("Date parse fail:", work["published_at_utc"].isna().sum())
print("Date range:", work["published_at_utc"].min(), "->", work["published_at_utc"].max())
print(work["published_year"].value_counts().sort_index().to_string())
print("title_too_short:", work["title_too_short"].sum())
print("content_too_short:", work["content_too_short"].sum())
print("possible_block_or_paywall:", work["possible_block_or_paywall"].sum())
print("content_word_count max:", work["content_word_count"].max())
```

```python
puzzle_pat = r"Wordle|Connections|Strands|Mini Crossword|Spelling Bee"
work["is_puzzle_template"] = work["title_clean"].fillna("").astype(str).str.contains(puzzle_pat, case=False, regex=True)
work["is_too_long"] = work["content_word_count"] > 5000

work["is_duplicate_url"] = work["url_valid"] & work["url"].duplicated(keep="first")
sig = (work["title_clean"].fillna("").astype(str).str.casefold() + "||" +
       work["content_clean"].fillna("").astype(str).str.casefold())
work["is_duplicate_text"] = sig.duplicated(keep="first")
work["is_duplicate_content"] = work["content_clean"].notna() & work["content_clean"].duplicated(keep="first")
work["is_duplicate_title"] = work["title_clean"].notna() & work["title_clean"].duplicated(keep="first")

work["missing_url"] = ~work["url_valid"]
work["missing_title"] = work["title_clean"].isna()
work["missing_content"] = work["content_clean"].isna()
work["missing_time"] = work["published_at_utc"].isna()
work["missing_author"] = work["author_name_clean"].isna()

rej = pd.DataFrame({
    "missing_or_invalid_url": work["missing_url"],
    "missing_title": work["missing_title"],
    "missing_content": work["missing_content"],
    "duplicate_url": work["is_duplicate_url"],
    "duplicate_content": work["is_duplicate_content"],
    "puzzle_template": work["is_puzzle_template"],
    "too_long_5000_words": work["is_too_long"],
})
work["reject_reason"] = rej.apply(lambda r: ";".join(r.index[r.values]), axis=1)
work["is_rejected"] = work["reject_reason"] != ""

rev = pd.DataFrame({
    "duplicate_title": work["is_duplicate_title"],
    "short_content": work["content_too_short"] & ~work["missing_content"],
    "short_title": work["title_too_short"] & ~work["missing_title"],
    "missing_or_invalid_time": work["missing_time"],
    "possible_block_or_paywall": work["possible_block_or_paywall"],
    "future_publication_date": work["published_at_utc"] > pd.Timestamp("2026-10-29", tz="UTC"),
})
work["review_reason"] = rev.apply(lambda r: ";".join(r.index[r.values]), axis=1)
work["needs_review"] = work["review_reason"] != ""

rejected_df = work[work["is_rejected"]].copy()
clean_df = work[~work["is_rejected"]].copy()
review_df = clean_df[clean_df["needs_review"]].copy()

print("Raw      :", len(work))
print("Clean    :", len(clean_df))
print("Rejected :", len(rejected_df))
print("Review (clean ke andar):", len(review_df))
assert len(clean_df) + len(rejected_df) == len(work) == len(df_raw)

print("\nReject reasons:")
print(rejected_df["reject_reason"].str.split(";").explode().value_counts().to_string())
print("\nReview reasons:")
print(review_df["review_reason"].str.split(";").explode().value_counts().to_string())
```
