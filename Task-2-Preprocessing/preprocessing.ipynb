```python
import re

clean = df.copy()
report = [("Shuru", len(clean))]

def step(name, mask_keep):
    global clean
    before = len(clean)
    clean = clean[mask_keep(clean)].copy()
    report.append((name, before - len(clean)))

# 1. Duplicate content, phir duplicate title (pehli row rakho)
step("Duplicate content hataye", lambda d: ~d.duplicated(subset="content", keep="first"))
step("Duplicate title hataye",   lambda d: ~d.duplicated(subset="title", keep="first"))

# 2. Puzzle-hint articles
puzzle_pat = r"Wordle|Connections|Strands|Mini Crossword|Spelling Bee"
step("Puzzle articles hataye", lambda d: ~d["title"].str.contains(puzzle_pat, case=False, regex=True))

# 3. 5000+ words (live blogs, transcripts, deal trackers)
step("5000+ words hataye", lambda d: d["content_words"] <= 5000)

# 4. Titles ke extra spaces
clean["title"] = clean["title"].str.replace(r"\s+", " ", regex=True).str.strip()

# 5. content_clean: HTML, URLs, emails, extra spaces
def clean_text(t):
    t = re.sub(r"<[^>]+>", " ", t)
    t = re.sub(r"http\S+|www\.\S+", " ", t)
    t = re.sub(r"\S+@\S+\.\S+", " ", t)
    t = re.sub(r"&nbsp;|&amp;", " ", t)
    return re.sub(r"\s+", " ", t).strip()

clean["content_clean"] = clean["content"].apply(clean_text)
clean["content_words"] = clean["content_clean"].str.split().str.len()

# 6. Safai ke baad jo bohat chhote reh gaye
step("Safai ke baad 50 words se kam hataye", lambda d: d["content_words"] >= 50)

clean = clean.reset_index(drop=True)

print("Cleaning report:")
for name, n in report:
    print(f"  {name}: {n}")
print("Bache:", len(clean))
print(clean["source"].value_counts().to_string())
```

```python
MISSING_TOKENS = {"", "n/a", "na", "null", "none", "nan", "-", "--"}

def normalize_missing(series, extra=()):
    s = series.astype(object)
    def f(v):
        if pd.isna(v):
            return pd.NA
        if isinstance(v, str):
            t = v.replace("\u00a0", " ").strip()
            if t.casefold() in MISSING_TOKENS or t.casefold() in extra:
                return pd.NA
            return t
        return v
    return s.map(f)

before = work[["title", "description", "author_name", "content", "url"]].isna().sum()

for c in ["title", "description", "content", "url", "date_of_publication"]:
    work[c] = normalize_missing(work[c])

# "Unknown" sirf author mein missing maana jayega
work["author_name"] = normalize_missing(work["author_name"], extra={"unknown"})

after = work[["title", "description", "author_name", "content", "url"]].isna().sum()
print(pd.DataFrame({"null_before": before, "null_after": after}))
print("\nauthor_name 'Unknown' bacha:", (work["author_name"] == "Unknown").sum())
```

```python
from html.parser import HTMLParser

class TextExtractor(HTMLParser):
    SKIP = {"script", "style", "noscript"}
    def __init__(self):
        super().__init__(convert_charrefs=True)
        self.parts, self._skip = [], 0
    def handle_starttag(self, tag, attrs):
        if tag in self.SKIP:
            self._skip += 1
    def handle_endtag(self, tag):
        if tag in self.SKIP and self._skip:
            self._skip -= 1
    def handle_data(self, data):
        if self._skip == 0 and data and data.strip():
            self.parts.append(data.strip())

INVISIBLE = dict.fromkeys(map(ord, "\u200b\u200c\u200d\u2060\ufeff\u00ad"), None)
HAS_TAG = re.compile(r"<[a-zA-Z/!][^>]*>")

def clean_text(v):
    if pd.isna(v):
        return pd.NA
    t = str(v)
    if HAS_TAG.search(t):                      # parser sirf tab jab HTML ho (tez)
        p = TextExtractor()
        try:
            p.feed(t); p.close()
            t = " ".join(p.parts)
        except Exception:
            t = re.sub(r"<[^>]+>", " ", t)
    t = html.unescape(t)
    t = t.translate(INVISIBLE)
    t = unicodedata.normalize("NFKC", t)
    t = (t.replace("\u201c", '"').replace("\u201d", '"')
          .replace("\u2018", "'").replace("\u2019", "'"))
    t = re.sub(r"\s+", " ", t).strip()
    return t if t else pd.NA

for c in ["title", "description", "author_name", "content"]:
    work[c + "_clean"] = work[c].map(clean_text)

print(work[["title_clean", "description_clean", "author_name_clean", "content_clean"]].isna().sum())

changed = (work["content"].astype(str) != work["content_clean"].astype(str)).sum()
print("\ncontent jis mein tabdili hui:", changed)
print("content_clean mein new line:", work["content_clean"].astype(str).str.contains("\n").sum())
print("description_clean mein new line:", work["description_clean"].astype(str).str.contains("\n").sum())
```

```python
df = data.copy().reset_index(drop=True)
df["row_index"] = df.index
issues = []

def add_issue(rule_id, severity, row_index, field, message):
    issues.append({"rule_id": rule_id, "severity": severity, "row_index": int(row_index),
                   "field": field, "message": message})

def is_valid_http_url(v):
    if not isinstance(v, str) or not v.strip():
        return False
    try:
        p = urlparse(v.strip())
        return p.scheme.lower() in {"http", "https"} and bool(p.hostname) and "." in p.hostname
    except Exception:
        return False

for col in CONTRACT["required_columns"]:
    if col not in df.columns:
        raise ValueError(f"Missing required column: {col}")

# required fields
for field in CONTRACT["required_columns"]:
    bad = df[field].isna() | (df[field].astype(str).str.strip() == "")
    for i in df.index[bad]:
        add_issue("FIELD_001", "ERROR", i, field, f"Required field '{field}' is missing or blank.")

# URL
for i in df.index[~df["url"].map(is_valid_http_url)]:
    add_issue("URL_001", "ERROR", i, "url", "URL is not a plausible HTTP(S) URL.")

# dates
pub = pd.to_datetime(df["date_of_publication"], errors="coerce", utc=True, format="mixed")
for i in df.index[df["date_of_publication"].notna() & pub.isna()]:
    add_issue("TIME_002", "WARNING", i, "date_of_publication", "Timestamp cannot be parsed.")
for i in df.index[pub >= pd.Timestamp("2026-10-31", tz="UTC")]:
    add_issue("TIME_003", "WARNING", i, "date_of_publication", "Publication date is in the future.")

# text
tl = df["title"].fillna("").astype(str).str.len()
cl = df["content"].fillna("").astype(str).str.len()
for i in df.index[(tl > 0) & (tl < CONTRACT["min_title_chars_warning"])]:
    add_issue("TEXT_001", "WARNING", i, "title", "Title is very short.")
for i in df.index[(cl > 0) & (cl < CONTRACT["min_content_chars_warning"])]:
    add_issue("TEXT_002", "WARNING", i, "content", "Content is very short.")
err = re.compile("|".join(CONTRACT["known_error_patterns"]), re.I)
blocked = df["content"].fillna("").astype(str).str.contains(err) & (cl < CONTRACT["block_page_max_chars"])
for i in df.index[blocked]:
    add_issue("TEXT_003", "WARNING", i, "content", "Content may be a block/CAPTCHA/error page.")

# duplicates
for i in df.index[df["url"].duplicated(keep=False)]:
    add_issue("DUP_001", "WARNING", i, "url", "URL occurs more than once.")
for i in df.index[df["content"].duplicated(keep=False)]:
    add_issue("DUP_002", "WARNING", i, "content", "Content occurs more than once.")
for i in df.index[df["title"].duplicated(keep=False)]:
    add_issue("DUP_003", "WARNING", i, "title", "Title occurs more than once.")

issues_df = pd.DataFrame(issues, columns=["rule_id", "severity", "row_index", "field", "message"])
print("Total issues:", len(issues_df))
print(issues_df.groupby(["rule_id", "severity"]).size().to_string() if len(issues_df) else "No issues: warnings 0, errors 0")
```

```python
import glob, re, html, json, unicodedata, os, math
from urllib.parse import urlparse
from html.parser import HTMLParser
import numpy as np
import pandas as pd

# ---------- 1. Raw load ----------
RAW_GLOB = "../Task-1-Web-Scraping-100k/dataset/*.zip"
df_raw = pd.concat([pd.read_csv(f) for f in sorted(glob.glob(RAW_GLOB))], ignore_index=True)
expected = ["title", "description", "date_of_publication", "author_name", "content", "url", "source"]
work = df_raw.copy(deep=True)
work["row_id"] = range(len(work))
for c in expected:
    work[c + "_raw"] = work[c]
print("1. Raw:", work.shape)

# ---------- 2. Missing tokens ----------
MISSING_TOKENS = {"", "n/a", "na", "null", "none", "nan", "-", "--"}

def normalize_missing(series, extra=()):
    def f(v):
        if pd.isna(v):
            return pd.NA
        if isinstance(v, str):
            t = v.replace("\u00a0", " ").strip()
            if t.casefold() in MISSING_TOKENS or t.casefold() in extra:
                return pd.NA
            return t
        return v
    return series.astype(object).map(f)

for c in ["title", "description", "content", "url", "date_of_publication"]:
    work[c] = normalize_missing(work[c])
work["author_name"] = normalize_missing(work["author_name"], extra={"unknown"})
print("2. author NULL:", work["author_name"].isna().sum())

# ---------- 3. Text cleaning ----------
class TextExtractor(HTMLParser):
    SKIP = {"script", "style", "noscript"}
    def __init__(self):
        super().__init__(convert_charrefs=True)
        self.parts, self._skip = [], 0
    def handle_starttag(self, tag, attrs):
        if tag in self.SKIP:
            self._skip += 1
    def handle_endtag(self, tag):
        if tag in self.SKIP and self._skip:
            self._skip -= 1
    def handle_data(self, data):
        if self._skip == 0 and data and data.strip():
            self.parts.append(data.strip())

INVISIBLE = dict.fromkeys(map(ord, "\u200b\u200c\u200d\u2060\ufeff\u00ad"), None)
HAS_TAG = re.compile(r"<[a-zA-Z/!][^>]*>")

def clean_text(v):
    if pd.isna(v):
        return pd.NA
    t = str(v)
    if HAS_TAG.search(t):
        p = TextExtractor()
        try:
            p.feed(t); p.close()
            t = " ".join(p.parts)
        except Exception:
            t = re.sub(r"<[^>]+>", " ", t)
    t = html.unescape(t)
    t = t.translate(INVISIBLE)
    t = unicodedata.normalize("NFKC", t)
    t = (t.replace("\u201c", '"').replace("\u201d", '"')
          .replace("\u2018", "'").replace("\u2019", "'"))
    t = re.sub(r"\s+", " ", t).strip()
    return t if t else pd.NA

for c in ["title", "description", "author_name", "content"]:
    work[c + "_clean"] = work[c].map(clean_text)
print("3. text cleaned")

# ---------- 4. URL, dates, length, block check ----------
def valid_url(v):
    if pd.isna(v):
        return False
    try:
        p = urlparse(str(v).strip())
        return p.scheme.lower() in {"http", "https"} and bool(p.hostname) and "." in p.hostname
    except Exception:
        return False

work["url_valid"] = work["url"].map(valid_url)
work["published_at_utc"] = pd.to_datetime(work["date_of_publication"], errors="coerce", utc=True, format="mixed")
work["title_char_count"] = work["title_clean"].fillna("").astype(str).str.len()
work["content_char_count"] = work["content_clean"].fillna("").astype(str).str.len()
work["content_word_count"] = work["content_clean"].fillna("").astype(str).str.findall(r"\b[\w'-]+\b").str.len()
work["title_too_short"] = work["title_char_count"] < 3
work["content_too_short"] = work["content_char_count"] < 200
block = re.compile(r"access denied|enable javascript|captcha|robot check|page not found|403 forbidden|subscribe to continue", re.I)
has_kw = work["content_clean"].fillna("").astype(str).str.contains(block)
work["possible_block_or_paywall"] = has_kw & (work["content_char_count"] < 500)
print("4. url/dates/length done")

# ---------- 5. Flags, rules, split ----------
puzzle_pat = r"Wordle|Connections|Strands|Mini Crossword|Spelling Bee"
work["is_puzzle_template"] = work["title_clean"].fillna("").astype(str).str.contains(puzzle_pat, case=False, regex=True)
work["is_too_long"] = work["content_word_count"] > 5000
work["is_duplicate_url"] = work["url_valid"] & work["url"].duplicated(keep="first")
work["is_duplicate_content"] = work["content_clean"].notna() & work["content_clean"].duplicated(keep="first")
work["missing_url"] = ~work["url_valid"]
work["missing_title"] = work["title_clean"].isna()
work["missing_content"] = work["content_clean"].isna()
work["missing_time"] = work["published_at_utc"].isna()

rej = pd.DataFrame({
    "missing_or_invalid_url": work["missing_url"],
    "missing_title": work["missing_title"],
    "missing_content": work["missing_content"],
    "duplicate_url": work["is_duplicate_url"],
    "duplicate_content": work["is_duplicate_content"],
    "puzzle_template": work["is_puzzle_template"],
    "too_long_5000_words": work["is_too_long"],
})
base_rej = rej.any(axis=1)
dup_title = pd.Series(False, index=work.index)
keep = work[~base_rej & work["title_clean"].notna()]
dup_title.loc[keep.index] = keep["title_clean"].duplicated(keep="first")
work["is_duplicate_title"] = dup_title

rej2 = rej.copy()
rej2["duplicate_title"] = work["is_duplicate_title"]
work["reject_reason"] = rej2.apply(lambda r: ";".join(r.index[r.values]), axis=1)
work["is_rejected"] = work["reject_reason"] != ""

rejected_df = work[work["is_rejected"]].copy()
clean_df = work[~work["is_rejected"]].copy()
assert len(clean_df) + len(rejected_df) == len(work) == len(df_raw)
print("5. Raw:", len(work), "| Clean:", len(clean_df), "| Rejected:", len(rejected_df))
```

```python
import glob, json, re, os
from urllib.parse import urlparse
import numpy as np
import pandas as pd

data = pd.concat(
    [pd.read_csv(f) for f in sorted(glob.glob("output/articles_clean_part*.csv.zip"))],
    ignore_index=True,
)
print("Shape:", data.shape)

CONTRACT = {
    "required_columns": ["url", "title", "content", "source"],
    "optional_columns": ["date_of_publication", "author_name", "description"],
    "min_title_chars_warning": 3,
    "min_content_chars_warning": 200,
    "block_page_max_chars": 500,
    "known_error_patterns": [r"access denied", r"enable javascript", r"captcha",
                             r"robot check", r"403 forbidden", r"page not found"],
}
print(json.dumps(CONTRACT, indent=2))
```

```python
df = data.copy().reset_index(drop=True)
df["row_index"] = df.index
issues = []

def add_issue(rule_id, severity, row_index, field, message):
    issues.append({"rule_id": rule_id, "severity": severity, "row_index": int(row_index),
                   "field": field, "message": message})

def is_valid_http_url(v):
    if not isinstance(v, str) or not v.strip():
        return False
    try:
        p = urlparse(v.strip())
        return p.scheme.lower() in {"http", "https"} and bool(p.hostname) and "." in p.hostname
    except Exception:
        return False

for col in CONTRACT["required_columns"]:
    if col not in df.columns:
        raise ValueError(f"Missing required column: {col}")

# required fields
for field in CONTRACT["required_columns"]:
    bad = df[field].isna() | (df[field].astype(str).str.strip() == "")
    for i in df.index[bad]:
        add_issue("FIELD_001", "ERROR", i, field, f"Required field '{field}' is missing or blank.")

# URL
for i in df.index[~df["url"].map(is_valid_http_url)]:
    add_issue("URL_001", "ERROR", i, "url", "URL is not a plausible HTTP(S) URL.")

# dates
pub = pd.to_datetime(df["date_of_publication"], errors="coerce", utc=True, format="mixed")
for i in df.index[df["date_of_publication"].notna() & pub.isna()]:
    add_issue("TIME_002", "WARNING", i, "date_of_publication", "Timestamp cannot be parsed.")
for i in df.index[pub >= pd.Timestamp("2026-10-31", tz="UTC")]:
    add_issue("TIME_003", "WARNING", i, "date_of_publication", "Publication date is in the future.")

# text
tl = df["title"].fillna("").astype(str).str.len()
cl = df["content"].fillna("").astype(str).str.len()
for i in df.index[(tl > 0) & (tl < CONTRACT["min_title_chars_warning"])]:
    add_issue("TEXT_001", "WARNING", i, "title", "Title is very short.")
for i in df.index[(cl > 0) & (cl < CONTRACT["min_content_chars_warning"])]:
    add_issue("TEXT_002", "WARNING", i, "content", "Content is very short.")
err = re.compile("|".join(CONTRACT["known_error_patterns"]), re.I)
blocked = df["content"].fillna("").astype(str).str.contains(err) & (cl < CONTRACT["block_page_max_chars"])
for i in df.index[blocked]:
    add_issue("TEXT_003", "WARNING", i, "content", "Content may be a block/CAPTCHA/error page.")

# duplicates
for i in df.index[df["url"].duplicated(keep=False)]:
    add_issue("DUP_001", "WARNING", i, "url", "URL occurs more than once.")
for i in df.index[df["content"].duplicated(keep=False)]:
    add_issue("DUP_002", "WARNING", i, "content", "Content occurs more than once.")
for i in df.index[df["title"].duplicated(keep=False)]:
    add_issue("DUP_003", "WARNING", i, "title", "Title occurs more than once.")

issues_df = pd.DataFrame(issues, columns=["rule_id", "severity", "row_index", "field", "message"])
print("Total issues:", len(issues_df))
print(issues_df.groupby(["rule_id", "severity"]).size().to_string() if len(issues_df) else "No issues: warnings 0, errors 0")
```

```python
FATAL_RULES = {"FIELD_001", "URL_001"}
row_rules = issues_df.groupby("row_index")["rule_id"].apply(list).to_dict()

def disposition(i):
    rules = row_rules.get(i, [])
    if any(r in FATAL_RULES for r in rules):
        return "QUARANTINE"
    return "WARNING" if rules else "ACCEPT"

df["validation_issue_codes"] = df["row_index"].map(lambda i: ";".join(row_rules.get(i, [])))
df["validation_disposition"] = df["row_index"].map(disposition)
accepted_df = df[df["validation_disposition"] == "ACCEPT"]
warning_df = df[df["validation_disposition"] == "WARNING"]
quarantine_df = df[df["validation_disposition"] == "QUARANTINE"]
print({"total": len(df), "accepted": len(accepted_df),
       "warning": len(warning_df), "quarantine": len(quarantine_df)})

def test_valid_url():
    assert is_valid_http_url("https://example.com/story/1")
    assert not is_valid_http_url("not-a-url")
    assert not is_valid_http_url("ftp://example.com/file")
    assert not is_valid_http_url(None)

def test_timestamp():
    assert pd.notna(pd.to_datetime("2026-10-27T09:30:00Z", errors="coerce", utc=True))
    assert pd.isna(pd.to_datetime("yesterday maybe", errors="coerce", utc=True))

def test_fully_clean():
    assert len(quarantine_df) == 0 and len(warning_df) == 0
    assert df["url"].is_unique and df["content"].is_unique and df["title"].is_unique

test_valid_url(); test_timestamp(); test_fully_clean()
print("All tests passed.")

summary = {
    "input_rows": int(len(df)),
    "total_issues": int(len(issues_df)),
    "issues_by_rule": {k: int(v) for k, v in issues_df.groupby("rule_id").size().items()},
    "dispositions": {"accept": len(accepted_df), "warning": len(warning_df),
                     "quarantine": len(quarantine_df)},
}
os.makedirs("output/validation", exist_ok=True)
issues_df.to_csv("output/validation/news_validation_issues.csv", index=False, encoding="utf-8-sig")
warning_df[["row_index", "source", "url", "title", "validation_issue_codes"]].to_csv(
    "output/validation/news_warning_review.csv", index=False, encoding="utf-8-sig")
quarantine_df[["row_index", "source", "url", "title", "validation_issue_codes"]].to_csv(
    "output/validation/news_quarantine.csv", index=False, encoding="utf-8-sig")
with open("output/validation/news_validation_summary.json", "w", encoding="utf-8") as f:
    json.dump(summary, f, indent=2)
print(json.dumps(summary, indent=2))
```
